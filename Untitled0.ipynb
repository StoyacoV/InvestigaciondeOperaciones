```python
import networkx as nx
import matplotlib.pyplot as plt

G=nx.Graph()
G.add_nodes_from([1,2,3,4,5,6])

list = [(1,2), (1,4), (1,5), (2,4), (2,6), (3,4), (5,6)]

G.add_edges_from(list)

nx.draw(G, with_labels=True, node_color="#F4ABAA", node_size=600)
```
