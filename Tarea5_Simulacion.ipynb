## Ejercicio 1: La rata en el laberinto (simulación Monte Carlo)

**Idea:** en vez de resolver las ecuaciones, repetimos el experimento muchas veces
con la computadora y vemos en qué proporción la rata llega a la comida (7)
antes que a la descarga (8).

**Paso a paso:**
1. Definimos el laberinto como un diccionario: para cada celda, la lista de celdas
   adyacentes a las que puede ir.
2. Simulamos UNA rata: empieza en la celda 0 y en cada paso elige una salida al azar
   (cada una con probabilidad 1/k) hasta llegar a 7 o a 8.
3. Repetimos N veces.
4. La probabilidad estimada = (veces que llegó a la comida) / N.

```python
import random

laberinto = {
    0: [2],
    1: [3],
    2: [0, 3, 8],
    3: [1, 2, 4, 5],
    4: [3, 6, 7],
    5: [3],
    6: [4],
}
COMIDA, DESCARGA = 7, 8
```

### Paso 2: simular una sola rata
La función devuelve True si la rata encontró la comida antes de recibir la descarga.
`random.choice` elige una salida con la misma probabilidad (1/k) para cada una.

```python
def simular_rata(inicio=0):
    celda = inicio
    while celda != COMIDA and celda != DESCARGA:
        celda = random.choice(laberinto[celda])
    return celda == COMIDA
```

### Paso 3 y 4: repetir N veces y estimar la probabilidad
Por la ley de los grandes números, mientras más repeticiones, más se acerca
la estimación al valor real (1/4 = 0.25).

```python
random.seed(42)
N = 200_000

exitos = sum(simular_rata(0) for _ in range(N))
prob = exitos / N

print(f"Veces que llegó a la comida: {exitos} de {N}")
print(f"Probabilidad estimada: {prob:.4f}")
print("Valor exacto (con matrices): 0.2500")
```

### Resultado Final
Obtenemos las iteraciones hasta aproximar el resultado (0.25).

```python
import random
lab = {0:[2], 1:[3], 2:[0,3,8], 3:[1,2,4,5], 4:[3,6,7], 5:[3], 6:[4]}

def rata():
    x = 0
    while x not in (7, 8):
        x = random.choice(lab[x])
    return x == 7

N = 800000
print("P(comida antes que descarga) ≈", sum(rata() for _ in range(N)) / N)
# Resultado esperado: ≈ 0.25
```

## Ejercicio 2: Serpientes y escaleras (simulación Monte Carlo)

**Idea:** jugamos muchísimas partidas solitarias con un dado y promediamos
cuántas tiradas tomó terminar cada una. Ese promedio estima el número esperado
de tiradas.

**Paso a paso:**
1. Definimos los atajos del tablero:
   - Escaleras: 3 → 11 y 15 → 19
   - Serpientes: 17 → 10 y 13 → 4
2. Una partida empieza en la casilla 1 con 0 tiradas.
3. En cada tirada: se lanza un dado (1 a 6), se avanza y se cuenta la tirada.
   Si se llega a 20 o más, el juego termina. Si no, y la casilla tiene escalera
   o serpiente, se mueve a donde indique.
4. Repetimos N partidas y calculamos el promedio de tiradas.

```python
import random

atajos = {
    3: 11,    # escalera
    15: 19,   # escalera
    17: 10,   # serpiente
    13: 4,    # serpiente
}
META = 20
```

### Paso 2 y 3: simular UNA partida
Devuelve el número de tiradas que se necesitaron para terminar.
Si el dado se pasa de 20, también se considera que el juego terminó.

```python
def simular_partida():
    casilla = 1
    tiradas = 0
    while casilla < META:
        casilla += random.randint(1, 6)   # tirar el dado
        tiradas += 1
        if casilla >= META:               # terminó el juego
            break
        casilla = atajos.get(casilla, casilla)  # aplicar escalera/serpiente si hay
    return tiradas
```

### Paso 4: repetir N partidas y promediar
Además del promedio, guardamos todas las duraciones para ver su distribución.

```python
random.seed(42)
N = 200_000

duraciones = [simular_partida() for _ in range(N)]
promedio = sum(duraciones) / N

print(f"Número esperado de tiradas (simulación): {promedio:.4f}")
print("Valor exacto (con matrices): 6.8983")
print(f"Partida más corta: {min(duraciones)}  |  más larga: {max(duraciones)}")
```

### Respuesta Final
Iteraciones hasta aproximar el resultado analítico (6.9)

```python
import random
atajos = {3: 11, 15: 19, 17: 10, 13: 4}

def partida():
    c, t = 1, 0
    while c < 20:
        c += random.randint(1, 6)
        t += 1
        if c >= 20:
            break
        c = atajos.get(c, c)
    return t

N = 200000
print("Tiros esperadis ≈", sum(partida() for _ in range(N)) / N)
# Resultado esperado: ≈ 6.90
```
